Problema de probabilidad condicional
En una fábrica se producen piezas mediante dos máquinas, A y B.

La máquina A produce el 60% del total de las piezas.
La máquina B produce el 40% restante.
Sin embargo, no son igual de precisas:

La máquina A produce 2% de piezas defectuosas.
La máquina B produce 5% de piezas defectuosas.
Un inspector toma una pieza al azar del inventario general y encuentra que la pieza está defectuosa.

Pregunta:

¿Cuál es la probabilidad de que la pieza haya sido fabricada por la máquina B?

```python
import numpy as np
import pandas as pd

import numpy as np
```

```python
p_maquina_a = 0.60
p_maquina_b = 0.40

p_defectuosas_a = 0.02
p_defectuosas_b = 0.05

# 1. P(D) = P(D|A)*P(A) + P(D|B)*P(B)
p_D = (p_defectuosas_a * p_maquina_a) + (p_defectuosas_b * p_maquina_b)

# 2. P(D ∩ B) - Probabilidad de que la pieza sea defectuosa Y de la M!quina B
p_D_inter_B = p_defectuosas_b * p_maquina_b

# --- Aplicaci!n de la F!rmula de Probabilidad Condicional ---
# P(B|D) = P(D ∩ B) / P(D)
p_B_dado_D = p_D_inter_B / p_D

# --- Mostrar el Resultado ---
print(f"Producci!n M!quina A: {p_maquina_a * 100:.0f}%")
print(f"Producci!n M!quina B: {p_maquina_b * 100:.0f}%")
print(f"Tasa de defectos M!quina A: {p_defectuosas_a * 100:.0f}%")
print(f"Tasa de defectos M!quina B: {p_defectuosas_b * 100:.0f}%")
print(f"\nP(D) probabilidad total de pieza defectuosa: {p_D:.4f}")
print(f"P(D ∩ B) probabilidad de defectuosa Y de M!quina B: {p_D_inter_B:.4f}")
print(f"La probabilidad condicional P(B|D) es: {p_B_dado_D:.3f} ({p_B_dado_D * 100:.1f}%)")
```

Problema practico de RMSE

```python


import numpy as np
import pandas as pd

import numpy as np

predichos = np.array([2, 8, 4, 9])
observados = np.array([3, 7, 5, 10])

errores = observados - predichos
errores_cuadrados = errores**2

suma_errores_cuadrados = np.sum(errores_cuadrados)
mse = suma_errores_cuadrados / len(errores)
rmse = np.sqrt(mse)

print(rmse)
```

Pregunta 21- Sensores

```python
import pandas as pd
df = pd.read_csv('/content/drive/MyDrive/sensores.csv', usecols=['ts', 'temperatura', 'humedad'])
display(df.head())

# Convertir la columna ts a datetime con formato mixto
df['ts'] = pd.to_datetime(df['ts'], format='mixed', errors='coerce')

print(df.dtypes)
display(df.head())
```

```python
# Reemplazar valores faltantes en temperatura con la mediana
df['temperatura'] = df['temperatura'].fillna(df['temperatura'].median())

# Reemplazar valores faltantes en humedad con la media
df['humedad'] = df['humedad'].fillna(df['humedad'].mean())

print(df.isnull().sum())
display(df.head())
```

```python
# ¿Cuántos NaT quedaron en ts?
nat_count = df['ts'].isna().sum()
print(f'Cantidad de NaT en la columna ts: {nat_count}')
```

```python
# Valor final de la mediana de temperatura tras imputación (redondeado a 1 decimal)
mediana_final = df['temperatura'].median()
print(f'La mediana final de temperatura es: {round(mediana_final, 1)}')
```

```python
# Valor final de la media de humedad tras imputación (redondeado a 3 decimales)
media_final_hum = df['humedad'].mean()
print(f'La media final de humedad es: {round(media_final_hum, 3)}')
```

#Agua y Cafe

```python
import pandas as pd
import numpy as np


# Establece una semilla para reproducibilidad
np.random.seed(42)


# 1. Crea una lista de matrículas de estudiantes
student_ids = [f'{i:06d}' for i in range(172120, 172120 + 20)]


# 2. Asigna 10 estudiantes al grupo 'Café' y 10 al grupo 'Agua'
groups = ['Café'] * 10 + ['Agua'] * 10
np.random.shuffle(groups)


# 3. Simula calificacionen de examenes para el grupo café
# Mean=70, Std=5, asegurate de que los valores estén entre (50-100)
coffee_scores = np.random.normal(loc=70, scale=5, size=10).round(0)
coffee_scores = np.clip(coffee_scores, 50, 100).astype(int)


# 4. Simula calificaciones de examen para el grupo de 'Agua'
# Mean=75, Std=5, asegurate de que los valores estén entre (50-100)
water_scores = np.random.normal(loc=75, scale=5, size=10).round(0)
water_scores = np.clip(water_scores, 50, 100).astype(int)


# Combina los valores basado en el grupo
exam_scores = []
coffee_idx = 0
water_idx = 0
for group in groups:
    if group == 'Café':
        exam_scores.append(coffee_scores[coffee_idx])
        coffee_idx += 1
    else:
        exam_scores.append(water_scores[water_idx])
        water_idx += 1


# 5. Combina en un DataFrame de Pandas
df_students = pd.DataFrame({
    'Matrícula': student_ids,
    'Grupo': groups,
    'Calificaciones': exam_scores
})


print("Datos simulados:")
print(df_students.head())
```

```python
# Calcula la media de calificaciones para el grupo 'Agua'
calificaciones_agua = df_students[df_students['Grupo'] == 'Agua']['Calificaciones']
media_agua = calificaciones_agua.mean()

print(f"\nMedia de calificaciones del grupo 'Agua': {media_agua:.2f}")
```
