```python
# =============================================================
# Actividad: Clasificación de Dígitos Manuscritos con Minería de Datos
# Leo Martinez 177192
# =============================================================

# --- 1. Importaciones ---
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import load_digits
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
import seaborn as sns

# --- 2. Carga del dataset ---
digits = load_digits()
print(f"Forma de los datos: {digits.data.shape}")
print(f"Clases: {digits.target_names}")
print(f"Total de imágenes: {len(digits.data)}")

# --- 3. Visualización de imágenes ---
fig, axes = plt.subplots(2, 5, figsize=(10, 5))
for i, ax in enumerate(axes.flat):
    ax.imshow(digits.images[i], cmap='gray_r', interpolation='nearest')
    ax.set_title(f"Dígito: {digits.target[i]}")
    ax.axis('off')
plt.suptitle("Ejemplos del dataset Digits", fontsize=14)
plt.tight_layout()
plt.show()

# --- 4. División de datos (80% entrenamiento, 20% prueba) ---
X_train, X_test, y_train, y_test = train_test_split(
    digits.data, digits.target, test_size=0.2, random_state=42
)
print(f"Entrenamiento: {X_train.shape[0]} muestras")
print(f"Prueba: {X_test.shape[0]} muestras")

# =============================================================
# 5. Entrenamiento de los tres modelos
# =============================================================

# Modelo 1: K-Nearest Neighbors (k=5)
knn = KNeighborsClassifier(n_neighbors=5)
knn.fit(X_train, y_train)

# Modelo 2: Support Vector Machine (kernel RBF)
svm = SVC(kernel='rbf', gamma='scale')
svm.fit(X_train, y_train)

# Modelo 3: Random Forest (100 estimadores)
rf = RandomForestClassifier(n_estimators=100, random_state=42)
rf.fit(X_train, y_train)

# =============================================================
# 6. Evaluación de los modelos
# =============================================================

modelos = {"KNN (k=5)": knn, "SVM (RBF)": svm, "Random Forest": rf}
resultados = {}

for nombre, modelo in modelos.items():
    y_pred = modelo.predict(X_test)
    acc = accuracy_score(y_test, y_pred)
    resultados[nombre] = acc
    print(f"\n{'='*50}")
    print(f"Modelo: {nombre} — Accuracy: {acc:.4f}")
    print(f"{'='*50}")
    print(classification_report(y_test, y_pred))

# --- 7. Comparación de accuracy ---
print("\n" + "="*50)
print("COMPARACIÓN DE MODELOS")
print("="*50)
for nombre, acc in resultados.items():
    print(f"  {nombre}: {acc:.4f} ({acc*100:.2f}%)")
mejor = max(resultados, key=resultados.get)
print(f"\n  ★ Mejor modelo: {mejor} con {resultados[mejor]*100:.2f}% de accuracy")

# --- 8. Gráfica comparativa de accuracy ---
plt.figure(figsize=(8, 5))
colores = ['#3498db', '#e74c3c', '#2ecc71']
barras = plt.bar(resultados.keys(), resultados.values(), color=colores)
for barra, acc in zip(barras, resultados.values()):
    plt.text(barra.get_x() + barra.get_width()/2, barra.get_height() + 0.002,
             f'{acc*100:.2f}%', ha='center', va='bottom', fontweight='bold')
plt.ylim(0.9, 1.01)
plt.ylabel("Accuracy")
plt.title("Comparación de Accuracy de los Modelos")
plt.tight_layout()
plt.show()

# --- 9. Matrices de confusión ---
fig, axes = plt.subplots(1, 3, figsize=(18, 5))
for ax, (nombre, modelo) in zip(axes, modelos.items()):
    y_pred = modelo.predict(X_test)
    mat = confusion_matrix(y_test, y_pred)
    sns.heatmap(mat, annot=True, fmt='d', cmap='Blues', ax=ax,
                xticklabels=digits.target_names, yticklabels=digits.target_names)
    ax.set_title(f"{nombre}")
    ax.set_xlabel("Predicho")
    ax.set_ylabel("Real")
plt.suptitle("Matrices de Confusión", fontsize=14)
plt.tight_layout()
plt.show()

# =============================================================
# 10. Creación de 3 imágenes manuscritas propias (8×8)
# =============================================================

# Imagen 1: Dígito "0"
img_0 = np.array([
    [ 0,  0,  5, 10, 10,  5,  0,  0],
    [ 0,  5, 12,  4,  4, 12,  5,  0],
    [ 0, 10,  8,  0,  0,  8, 10,  0],
    [ 0, 10,  4,  0,  0,  4, 10,  0],
    [ 0, 10,  4,  0,  0,  4, 10,  0],
    [ 0, 10,  8,  0,  0,  8, 10,  0],
    [ 0,  5, 12,  4,  4, 12,  5,  0],
    [ 0,  0,  5, 10, 10,  5,  0,  0],
], dtype=float)

# Imagen 2: Dígito "1"
img_1 = np.array([
    [ 0,  0,  0,  8,  4,  0,  0,  0],
    [ 0,  0,  6, 14,  4,  0,  0,  0],
    [ 0,  0,  2, 12,  4,  0,  0,  0],
    [ 0,  0,  0, 12,  4,  0,  0,  0],
    [ 0,  0,  0, 12,  4,  0,  0,  0],
    [ 0,  0,  0, 12,  4,  0,  0,  0],
    [ 0,  0,  4, 14, 12,  4,  0,  0],
    [ 0,  0,  2, 10, 10,  2,  0,  0],
], dtype=float)

# Imagen 3: Dígito "7"
img_7 = np.array([
    [ 0,  2, 10, 12, 12, 12,  8,  0],
    [ 0,  0,  2,  4,  4, 10,  6,  0],
    [ 0,  0,  0,  0,  4, 12,  2,  0],
    [ 0,  0,  0,  0,  8,  8,  0,  0],
    [ 0,  0,  0,  4, 12,  2,  0,  0],
    [ 0,  0,  0,  8, 10,  0,  0,  0],
    [ 0,  0,  2, 12,  4,  0,  0,  0],
    [ 0,  0,  4, 12,  2,  0,  0,  0],
], dtype=float)

# Etiquetas reales de las imágenes creadas
imagenes_propias = [img_0, img_1, img_7]
etiquetas_reales = [0, 1, 7]

# --- 11. Visualización de las imágenes creadas ---
fig, axes = plt.subplots(1, 3, figsize=(9, 3))
for ax, img, etiq in zip(axes, imagenes_propias, etiquetas_reales):
    ax.imshow(img, cmap='gray_r', interpolation='nearest')
    ax.set_title(f"Dígito real: {etiq}")
    ax.axis('off')
plt.suptitle("Imágenes manuscritas propias (8×8)", fontsize=14)
plt.tight_layout()
plt.show()

# =============================================================
# 12. Predicción con los tres modelos
# =============================================================

# Convertir cada imagen 8x8 a un vector de 64 valores
X_propias = np.array([img.flatten() for img in imagenes_propias])

print("\n" + "="*60)
print("PREDICCIÓN DE IMÁGENES PROPIAS CON LOS TRES MODELOS")
print("="*60)
print(f"{'Dígito Real':<14} {'KNN (k=5)':<12} {'SVM (RBF)':<12} {'Random Forest':<14}")
print("-"*60)

for i, etiq_real in enumerate(etiquetas_reales):
    pred_knn = knn.predict(X_propias[i].reshape(1, -1))[0]
    pred_svm = svm.predict(X_propias[i].reshape(1, -1))[0]
    pred_rf  = rf.predict(X_propias[i].reshape(1, -1))[0]
    print(f"     {etiq_real:<12} {pred_knn:<12} {pred_svm:<12} {pred_rf:<14}")

# --- 13. Visualización de predicciones ---
fig, axes = plt.subplots(3, 3, figsize=(10, 10))
for i, (img, etiq_real) in enumerate(zip(imagenes_propias, etiquetas_reales)):
    for j, (nombre, modelo) in enumerate(modelos.items()):
        pred = modelo.predict(img.flatten().reshape(1, -1))[0]
        axes[i][j].imshow(img, cmap='gray_r', interpolation='nearest')
        acierto = "✓" if pred == etiq_real else "✗"
        color = 'green' if pred == etiq_real else 'red'
        axes[i][j].set_title(f"{nombre}\nReal: {etiq_real} → Pred: {pred} {acierto}",
                              color=color, fontsize=10)
        axes[i][j].axis('off')
plt.suptitle("Predicciones de imágenes propias por cada modelo", fontsize=14)
plt.tight_layout()
plt.show()

# =============================================================
# 14. Conclusiones
# =============================================================
print("\n" + "="*60)
print("CONCLUSIONES")
print("="*60)
print(f"""
1. Se entrenaron tres modelos (KNN, SVM, Random Forest) sobre
   el dataset Digits de sklearn ({len(digits.data)} imágenes de 8×8).

2. Accuracy en conjunto de prueba:""")
for nombre, acc in resultados.items():
    print(f"   - {nombre}: {acc*100:.2f}%")
print(f"""
3. El mejor modelo fue: {mejor} con {resultados[mejor]*100:.2f}% de accuracy.

4. Se crearon 3 imágenes manuscritas propias (dígitos 0, 1 y 7)
   y se clasificaron con los tres modelos entrenados.

5. Los tres algoritmos mostraron un desempeño alto en el dataset
   Digits. SVM con kernel RBF tiende a ser el más preciso para
   este tipo de datos de alta dimensionalidad.
""")
```
